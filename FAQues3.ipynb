```python
import pandas as pd

df = pd.read_excel('/content/CF_data.xlsx', sheet_name='DebtForg')
df.head()
```

```python
df.info()
print(df.isnull().sum())
```

```python
debt = df["y2014"].dropna()

print("Mean:", debt.mean())
print("Median:", debt.median())
print("Variance:", debt.var())
print("Skewness:", debt.skew())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
plt.boxplot(debt)
plt.title("Debt-to-GDP Ratio (2014)")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.hist(debt, bins=20)
plt.title("Distribution of Debt-to-GDP Ratio (2014)")
plt.xlabel("Debt-to-GDP Ratio")
plt.ylabel("Number of Countries")
plt.show()
```

```python
import pandas as pd

debt = pd.read_excel('/content/CF_data.xlsx', sheet_name='DebtForg')
gdp = pd.read_excel('/content/CF_data.xlsx', sheet_name='GDP')
```

```python
merged = pd.merge(
    debt[['countrycode','country','y2014']],
    gdp[['countrycode','y2014']],
    on='countrycode',
    suffixes=('_Debt','_GDP')
)

merged.head()
```

```python
merged['Debt_to_GDP_Ratio'] = (
    merged['y2014_Debt'] / merged['y2014_GDP']
) * 100

merged[['country','Debt_to_GDP_Ratio']].head()
```

```python
ratio = merged['Debt_to_GDP_Ratio'].dropna()
```

```python
print("Mean:", ratio.mean())
print("Median:", ratio.median())
print("Variance:", ratio.var())
print("Skewness:", ratio.skew())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
plt.boxplot(ratio)
plt.title("Debt-to-GDP Ratio (2014)")
plt.ylabel("Debt-to-GDP Ratio (%)")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.hist(ratio, bins=20)
plt.title("Distribution of Debt-to-GDP Ratio (2014)")
plt.xlabel("Debt-to-GDP Ratio (%)")
plt.ylabel("Number of Countries")
plt.show()
```
